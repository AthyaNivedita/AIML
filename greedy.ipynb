```python
def greedy_first_fit(items, bins_capacity):
    # Initialize an empty list of bins
    bins = []

    # Iterate through each item to place it in a bin
    for item in items:
        placed = False
        # Try to place the item in an existing bin
        for i in range(len(bins)):
            if bins_capacity - sum(bins[i]) >= item:
                bins[i].append(item)
                placed = True
                break
        # If the item could not be placed in any existing bin, create a new bin
        if not placed:
            bins.append([item])

    return bins

# Example usage:
# items = [2, 5, 4, 7, 1, 3, 8]
# bins_capacity = 10
# result_bins = greedy_first_fit(items, bins_capacity)
# print(f"Items: {items}")
# print(f"Bin Capacity: {bins_capacity}")
# print(f"Bins after First Fit: {result_bins}")
# print(f"Number of bins used: {len(result_bins)}")
```

### Example Usage of Greedy First Fit

```python
items = [2, 5, 4, 7, 1, 3, 8]
bins_capacity = 10

result_bins = greedy_first_fit(items, bins_capacity)

print(f"Items to pack: {items}")
print(f"Bin Capacity: {bins_capacity}")
print(f"\nBins after First Fit:\n{result_bins}")
print(f"Number of bins used: {len(result_bins)}")
```

### Breadth-First Search (BFS) Implementation

```python
from collections import deque

def bfs(graph, start_node):
    visited = set()  # To keep track of visited nodes
    queue = deque([start_node])  # Initialize queue with the starting node
    visited.add(start_node)  # Mark the start node as visited
    traversal_order = []  # To store the order of visited nodes

    while queue:
        current_node = queue.popleft()  # Dequeue a node
        traversal_order.append(current_node) # Add to traversal order

        # Enqueue all unvisited neighbors of the current_node
        for neighbor in graph.get(current_node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return traversal_order

# Example Usage:
# Represent the graph using an adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

start_node = 'A'
bfs_result = bfs(graph, start_node)

print(f"Graph: {graph}")
print(f"Starting node for BFS: {start_node}")
print(f"BFS Traversal Order: {bfs_result}")
```

### Depth-First Search (DFS) Implementation

```python
def dfs(graph, start_node):
    visited = set()  # To keep track of visited nodes
    stack = [start_node]  # Initialize stack with the starting node
    traversal_order = []  # To store the order of visited nodes

    while stack:
        current_node = stack.pop()  # Pop a node from the stack

        if current_node not in visited:
            visited.add(current_node)  # Mark it as visited
            traversal_order.append(current_node)  # Add to traversal order

            # Push unvisited neighbors onto the stack
            # We iterate in reverse to get a consistent order for printing
            # as stack.pop() will retrieve the last added element.
            for neighbor in reversed(graph.get(current_node, [])):
                if neighbor not in visited:
                    stack.append(neighbor)
    return traversal_order

dfs_result = dfs(graph, start_node)

print(f"Graph: {graph}")
print(f"Starting node for DFS: {start_node}")
print(f"DFS Traversal Order: {dfs_result}")
```
