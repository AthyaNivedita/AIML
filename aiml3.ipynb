```python
import heapq

def uniform_cost_search(graph, start, goal):


    # Priority queue stores tuples of (cost, node, path)
    # The lowest cost item is always popped first
    priority_queue = [(0, start, [start])]

    # Set to keep track of visited nodes to avoid cycles and redundant processing
    visited = set()

    while priority_queue:
        current_cost, current_node, current_path = heapq.heappop(priority_queue)

        if current_node == goal:
            return current_path, current_cost

        if current_node in visited:
            continue

        visited.add(current_node)

        for neighbor, cost in graph.get(current_node, {}).items():
            if neighbor not in visited:
                new_cost = current_cost + cost
                new_path = current_path + [neighbor]
                heapq.heappush(priority_queue, (new_cost, neighbor, new_path))

    return None, float('inf')

# Example Usage:
graph_ucs = {
    'A': {'B': 1, 'C': 10},
    'B': {'D': 2},
    'C': {'D': 3, 'E': 4},
    'D': {'E': 5},
    'E': {}
}

start_node_ucs = 'A'
goal_node_ucs = 'E'

path_ucs, cost_ucs = uniform_cost_search(graph_ucs, start_node_ucs, goal_node_ucs)

if path_ucs:
    print(f"UCS Path from {start_node_ucs} to {goal_node_ucs}: {path_ucs}")
    print(f"UCS Cost: {cost_ucs}")
else:
    print(f"No path found from {start_node_ucs} to {goal_node_ucs}")
```

```python
import heapq

def a_star_search(graph, start, goal, heuristic):

    # Priority queue stores tuples of (f_cost, g_cost, node, path)
    # f_cost = g_cost + heuristic[node]
    priority_queue = [(heuristic[start], 0, start, [start])]

    # Dictionary to store the minimum cost to reach a node (g_cost)
    g_costs = {node: float('inf') for node in graph}
    g_costs[start] = 0

    # Set to keep track of visited nodes
    visited = set()

    while priority_queue:
        f_cost, current_g_cost, current_node, current_path = heapq.heappop(priority_queue)

        if current_node == goal:
            return current_path, current_g_cost

        if current_node in visited:
            continue

        visited.add(current_node)

        for neighbor, edge_cost in graph.get(current_node, {}).items():
            new_g_cost = current_g_cost + edge_cost

            if new_g_cost < g_costs.get(neighbor, float('inf')):
                g_costs[neighbor] = new_g_cost
                new_f_cost = new_g_cost + heuristic.get(neighbor, 0)
                new_path = current_path + [neighbor]
                heapq.heappush(priority_queue, (new_f_cost, new_g_cost, neighbor, new_path))

    return None, float('inf')

# Example Usage:
graph_a_star = {
    'A': {'B': 1, 'C': 4},
    'B': {'D': 2, 'E': 5},
    'C': {'F': 1},
    'D': {'G': 3},
    'E': {'G': 1},
    'F': {'G': 2},
    'G': {}
}

# Heuristic values (estimated cost from node to goal)
heuristic_a_star = {
    'A': 7,
    'B': 6,
    'C': 2,
    'D': 3,
    'E': 1,
    'F': 1,
    'G': 0
}

start_node_a_star = 'A'
goal_node_a_star = 'G'

path_a_star, cost_a_star = a_star_search(graph_a_star, start_node_a_star, goal_node_a_star, heuristic_a_star)

if path_a_star:
    print(f"A* Path from {start_node_a_star} to {goal_node_a_star}: {path_a_star}")
    print(f"A* Cost: {cost_a_star}")
else:
    print(f"No path found from {start_node_a_star} to {goal_node_a_star}")
```
